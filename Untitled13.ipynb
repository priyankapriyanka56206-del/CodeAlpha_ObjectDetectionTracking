```python
!pip install -q ultralytics
from google.colab import files
from ultralytics import YOLO
import cv2, base64
from IPython.display import HTML

# 1. Upload any short video (people, cars, etc.)
uploaded = files.upload()
video_path = next(iter(uploaded))

# 2. Load pre-trained YOLOv8 model
model = YOLO("yolov8n.pt")

# 3. Read video and prepare output
cap = cv2.VideoCapture(video_path)
w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS) or 25
out = cv2.VideoWriter("raw.mp4", cv2.VideoWriter_fourcc(*"mp4v"), fps, (w, h))

# 4. Detect + track each frame, draw boxes, labels and tracking IDs
while cap.isOpened():
    ok, frame = cap.read()
    if not ok:
        break
    result = model.track(frame, persist=True, tracker="bytetrack.yaml", verbose=False)[0]
    out.write(result.plot())   # boxes + class labels + track IDs
cap.release()
out.release()

# 5. Convert for browser playback, show it, and download it
!ffmpeg -y -loglevel error -i raw.mp4 -vcodec libx264 output.mp4
data = base64.b64encode(open("output.mp4", "rb").read()).decode()
display(HTML(f'<video width=640 controls><source src="data:video/mp4;base64,{data}"></video>'))
files.download("output.mp4")
```
